```python
# Step 1 : IMport required libraries

import numpy as np
import pandas as pd
```

```python
# Step 2 : load dataset

df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')

```

```python
# Step 3 : Explore the dataset

#  view first 5 records

df.head(5)

df.info()
```

```python
# Statistical summary

df.describe()
```

```python
# dataset shape

df.shape
```

```python
# Step 4 : check missing values

df.isnull().sum()
```

```python
# step 5 : check duplicate records

df.duplicated().sum()
```

```python
# remove duplicate 

df.drop_duplicates(inplace = True)

df.duplicated().sum()
```

```python
#  step 6 : replace invalid values
#  replace common incalid with nan

df.replace(['N/A','NULL',''],np.nan,inplace=True)

df
```

```python
# Step 7 : remove extra spaces

df = df.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
```

```python
# Step 8 : Standardize Text Format

#  Convert text into proper case.

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# Step  9 : Validate Email Addresses

#  Keep only valid Email records.

df = df[df['Email'].str.contains('@',na=False)]
```

```python
# Step 10 : Convert date columns 

df['Order_Date'] = pd.to_datetime(df['Order_Date'],errors='coerce')  # NAT
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# Step 11 : Convert Numeric columns 

cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
# Step 12 : Remove Invalid Quantity

#  remove records having negative or zero quantity

df = df[df['Qty']>0]
```

```python
# Step 13 : Fill missing values

# df['Discount'].fillna(0,inplace=True)

df['Discount'] = df['Discount'].fillna(0)

# df['Phone'].fillna('unknown', inplace = True)

df['Phone'] = df['Phone'].fillna('Unknown')

# df['Delivery_Date'].fillna('Delivery_Date', inplace = True)

df['Delivery_Date'] = df['Delivery_Date'].fillna('Order_Date')
```

```python
# Step 14: Feature Engineering

# Total Sales

df['Sales'] = df['Qty'] * df['Unit_Price']
```

```python
# Step 15  net amount 

df['Net_Amount'] = df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
# Profit (example ; 20%)

df['Profit'] = df['Net_Amount']*0.20
```
