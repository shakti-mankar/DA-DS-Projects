```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
%matplotlib
import seaborn as sns

```

```python
df =  pd.read_csv("Diwali Sales Data.csv" , encoding='unicode_escape')
```

```python
df
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
# drop unrelated blanks and columns

df.drop(['Status','unnamed1'], axis = 1 , inplace =True)
```

```python
df.info()
```

```python
# Check for null values

df.isnull().sum()
```

```python
# drop null values

df.dropna(inplace=True)
```

```python
# Change datatype

df['Amount'] = df['Amount'].astype('int')
```

```python
df['Amount'].astype
```

```python
df.columns
```

```python
#rename column
df.rename(columns= {'Marital_Status':'Shaadi'})
```

```python
# describe() method returns description of the data in the DataFrame (i.e. count, mean, std, etc)
df.describe()
```

```python
# use describe() for specific columns
df[['Age', 'Orders', 'Amount']].describe()
```

Exploratory Data analysis



```python
df.columns
```

```python
sns.countplot(x = 'Gender' , data = df)
```

```python
ax = sns.countplot(x = 'Gender', data =df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount' , ascending=False)
```

```python
sales_gender = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount' ,  ascending=False)

sns.barplot(x = 'Gender' , y = 'Amount' , data = sales_gender)
```

```python
# From the above graphs we can see that the most of the buyers are of age group between 26-35 years female
```

```python
# Total amount VS AGE group 

sales_age = df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Age Group' , y = 'Amount' , data = sales_age)
```

```python
#  Total number of orders from top 10 states

```

STATE

```python
df.columns
```

```python
# total number of orders from top 10 states

sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
```

```python
sales_state
```

```python
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State',y= 'Orders')
```

```python
# total number of Amount from top 10 states

sales_orders = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = sales_orders, x = 'State',y= 'Amount')
```


From above graphs we can see that most of the orders & total sales/amount are from Uttar Pradesh, Maharashtra and Karnataka respectively

```python
sales_state = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_ID',y= 'Orders')
```

```python
# top 10 most sold products (same thing as above)

fig1, ax1 = plt.subplots(figsize=(12,7))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False).plot(kind='bar')
```

Married women age group 26-35 yrs from UP, Maharastra and Karnataka working in IT, Healthcare and Aviation are more likely to buy products from Food, Clothing and Electronics category
